```python
import pandas as pd
from scipy import stats

df = pd.read_csv('user_metrics.csv', sep=';')

group_a = df[df['group'] == 'A']['revenue_total']
group_b = df[df['group'] == 'B']['revenue_total']

t_stat, p_value = stats.ttest_ind(group_a, group_b, equal_var=False)
print(f'Welch t-test: t={t_stat:.4f}, p-value={p_value:.6f}')

print(f'Среднее A: {group_a.mean():.2f}')
print(f'Среднее B: {group_b.mean():.2f}')

conv_a = df[df['group'] == 'A']['converted']
conv_b = df[df['group'] == 'B']['converted']

import numpy as np
p_a, p_b = conv_a.mean(), conv_b.mean()
n_a, n_b = len(conv_a), len(conv_a)

se = np.sqrt(p_a*(1-p_a)/n_a+p_b*(1-p_b)/n_b)
z = (p_b-p_a)/se
p_value_z = 2*(1-stats.norm.cdf(abs(z)))

print(f'Z-test: z={z:.4f}, p-value={p_value_z:.6f}')
print(f'Конверсия A: {p_a:.4f}, Конверсия B: {p_b:.4f}')
```
